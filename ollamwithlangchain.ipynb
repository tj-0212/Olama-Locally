```python
# Importing the necessary class from langchain_core.prompts
from langchain_core.prompts import ChatPromptTemplate

# Creating a ChatPromptTemplate object that defines the conversation structure.
chat_template = ChatPromptTemplate.from_messages(
    [
        # System message defines the role of the assistant
        ("system", "You are a helpful assistant that gives a one-line definition of the word entered by the user"),
        
        # Human message where the input from the user will be placed
        ("human", "{user_input}"),  # {user_input} is a placeholder to be replaced with the word the user enters
    ]
)


# The format_messages method will replace the placeholder {user_input} with the actual word passed here
messages = chat_template.format_messages(user_input="Sesquipedalian")

# Printing the formatted messages to check the output
print(messages)

```

```python
from langchain_ollama import ChatOllama  # Import the Ollama LLM wrapper

# Initialize the ChatOllama LLM.
llm = ChatOllama(
    model='llama3.2:latest',  # Specify the Ollama model to use.
    temperature=0  # Set the temperature (0 for deterministic output).
)

# You can test the LLM directly like this (before creating the chain).
# This is useful for isolating problems.
try:
    model_message = llm.invoke(messages)  # Pass the list of messages to the LLM
    print(model_message)
except Exception as e:
    print(f"Error invoking LLM directly: {e}")
```

```python
from langchain_core.output_parsers import StrOutputParser # For parsing LLM output to string
# Create the chain where input flows from chat template to llm and then to print out only the wnted content we use stroutparser
chain  = chat_template | llm | StrOutputParser()
```

```python
chain.invoke({"user_input":"genai"})
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain.embeddings import OllamaEmbeddings
from langchain_chroma import Chroma
from langchain_ollama import ChatOllama
```

```python
# Load the text file
raw_documents = TextLoader(r"C:\Users\Admin\Desktop\ollama\Langretrival.txt").load()
from langchain_text_splitters import RecursiveCharacterTextSplitter
# Split text into smaller chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=20)
documents = text_splitter.split_documents(raw_documents)

print(len(documents))

```

```python
oembeded = OllamaEmbeddings(base_url="http://localhost:11434", model="nomic-embed-text", show_progress=True)
```

```python
db = Chroma.from_documents(documents,embedding=oembeded)
```

```python
q = "Graph"
docs = db.similarity_search(q)
print(len(docs))
print(docs[0].page_content)
print(docs[1].page_content)
print(docs[2].page_content)
print(docs[3].page_content)
```

```python
from langchain_core.runnables import RunnablePassthrough 
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

template= """ Answer the question based on following context:
{context}

Question:{question}
"""
prompt = ChatPromptTemplate.from_template(template)
```

```python
model = ChatOllama(
    model = "llama3.2:latest",
    temperature=0
)
```

```python

retriver = db.as_retriever()
def format_input(docs):
    return "\n\n".join([d.page_content for d in docs])
```

```python
#RunnablePassthrough stores input in question 
#flow starts from retriver where inout is formated and then stores in context
#output is shown by StrOutputparser
chain = (
    {"context": retriver | format_input, "question": RunnablePassthrough()}
    | prompt
    | model
    | StrOutputParser()
)
```

```python
chain.invoke("tell me what you know about graph invocation")
```
